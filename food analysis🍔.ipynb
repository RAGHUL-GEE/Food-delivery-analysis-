```python
import pandas as pd
import numpy as np

df = pd.read_csv("https://drive.google.com/uc?export=download&id=1ln_YbLXHZjah8c-zgrT_zacnfSwTB9ZE", low_memory=False)


print("--- Missing Values Before Imputation ---")
print(df.isnull().sum()[df.isnull().sum() > 0])

df['Customer_Age'] = df['Customer_Age'].fillna(df['Customer_Age'].median())
df['Order_Value'] = df['Order_Value'].fillna(df['Order_Value'].median())
df['Final_Amount'] = df['Final_Amount'].fillna(df['Final_Amount'].median())
df['Discount_Applied'] = df['Discount_Applied'].fillna(df['Discount_Applied'].median())


df['Distance_km'] = df['Distance_km'].fillna(df['Distance_km'].mean())
df['Delivery_Time_Min'] = df['Delivery_Time_Min'].fillna(df['Delivery_Time_Min'].mean())


categorical_cols = ['Customer_Gender', 'City', 'Area', 'Cuisine_Type', 'Payment_Mode', 'Order_Status', 'Peak_Hour']
for col in categorical_cols:
    if col in df.columns and df[col].isnull().sum() > 0:
        mode_val = df[col].mode()[0]
        df[col] = df[col].fillna(mode_val)


print("\n--- Missing Values After Imputation ---")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
import pandas as pd
import numpy as np



def remove_outliers_iqr(data, columns):
    df_clean = data.copy()
    for col in columns:
        Q1 = df_clean[col].quantile(0.25)
        Q3 = df_clean[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        # Filter rows within bounds
        initial_count = len(df_clean)
        df_clean = df_clean[(df_clean[col] >= lower_bound) & (df_clean[col] <= upper_bound)]
        removed_count = initial_count - len(df_clean)
        print(f"Removed {removed_count:,} outliers from '{col}' using IQR method.")
        
    return df_clean


def cap_outliers_quantile(data, columns, lower_quantile=0.01, upper_quantile=0.99):
    df_capped = data.copy()
    for col in columns:
        lower_val = df_capped[col].quantile(lower_quantile)
        upper_val = df_capped[col].quantile(upper_quantile)
        
        
        df_capped[col] = df_capped[col].clip(lower=lower_val, upper=upper_val)
        print(f"Capped outliers in '{col}' between {lower_val:.2f} and {upper_val:.2f} ({lower_quantile*100}% - {upper_quantile*100}% percentiles).")
        
    return df_capped


df = cap_outliers_quantile(df, ['Delivery_Time_Min', 'Order_Value'], lower_quantile=0.01, upper_quantile=0.99)
```

```python
import pandas as pd
import numpy as np

rating_cols = ['Delivery_Rating', 'Restaurant_Rating']

for col in rating_cols:
    if col in df.columns:
        invalid_count = df[(df[col] > 5) | (df[col] < 1)][col].count()
        df[col] = df[col].clip(lower=1.0, upper=5.0)
        print(f"Corrected {invalid_count:,} invalid ratings in '{col}' (capped within range [1.0, 5.0]).")


invalid_margin_count = (df['Profit_Margin'] < 0).sum()
df['Profit_Margin'] = df['Profit_Margin'].clip(lower=0.0)
print(f"Corrected {invalid_margin_count:,} negative profit margin values (floored to 0.0).")
```

```python
import pandas as pd
import numpy as np

text_cols = ['Customer_Gender', 'City', 'Area', 'Cuisine_Type', 'Payment_Mode', 'Order_Status', 'Order_Day', 'Peak_Hour']

for col in text_cols:
    if col in df.columns:
        
        df[col] = df[col].astype(str).str.lower().str.strip()
       
        df[col] = df[col].replace('nan', np.nan)


if 'Customer_Gender' in df.columns:
    gender_mapping = {'m': 'male', 'f': 'female', 'male': 'male', 'female': 'female', 'other': 'other'}
    df['Customer_Gender'] = df['Customer_Gender'].map(gender_mapping).fillna(df['Customer_Gender'])


if 'City' in df.columns:
    df['City'] = df['City'].str.title() # Capitalizes first letter (e.g., 'chennai' -> 'Chennai')


if 'Order_Status' in df.columns:
    status_mapping = {'delivrd': 'delivered', 'canceld': 'cancelled', 'deliverd': 'delivered'}
    df['Order_Status'] = df['Order_Status'].replace(status_mapping)

print("Categorical standardization completed successfully.")
print("Unique Cities after standardization:", df['City'].unique() if 'City' in df.columns else "N/A")
print("Unique Order Statuses after standardization:", df['Order_Status'].unique() if 'Order_Status' in df.columns else "N/A")
```

```python
import pandas as pd
import numpy as np

if 'Order_Status' in df.columns:
    cancelled_mask = df['Order_Status'].astype(str).str.lower() == 'cancelled'

    
    if 'Delivery_Rating' in df.columns:
        invalid_del_ratings = df.loc[cancelled_mask, 'Delivery_Rating'].notnull().sum()
        df.loc[cancelled_mask, 'Delivery_Rating'] = np.nan
        print(f"Cleared {invalid_del_ratings:,} delivery ratings for cancelled orders.")

    if 'Restaurant_Rating' in df.columns:
        invalid_res_ratings = df.loc[cancelled_mask, 'Restaurant_Rating'].notnull().sum()
        df.loc[cancelled_mask, 'Restaurant_Rating'] = np.nan
        print(f"Cleared {invalid_res_ratings:,} restaurant ratings for cancelled orders.")


if 'Order_Status' in df.columns and 'Delivery_Time_Min' in df.columns:
    invalid_del_times = df.loc[cancelled_mask, 'Delivery_Time_Min'].notnull().sum()
    df.loc[cancelled_mask, 'Delivery_Time_Min'] = np.nan
    print(f"Cleared {invalid_del_times:,} delivery time values for cancelled orders.")


if all(col in df.columns for col in ['Order_Value', 'Discount_Applied', 'Final_Amount']):
    
    df['Final_Amount'] = df['Order_Value'] - df['Discount_Applied']
    
    df['Final_Amount'] = df['Final_Amount'].clip(lower=0.0)
    print("Recalculated 'Final_Amount' (Order_Value - Discount_Applied) across all records.")


if all(col in df.columns for col in ['Order_Status', 'Cancellation_Reason']):
    delivered_mask = df['Order_Status'].astype(str).str.lower() == 'delivered'
    df.loc[delivered_mask, 'Cancellation_Reason'] = 'none'
    print("Standardized 'Cancellation_Reason' to 'none' for all delivered orders.")

print("Logical consistency checks and corrections completed successfully.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

numeric_cols = [
    'Customer_Age', 
    'Delivery_Time_Min', 
    'Distance_km', 
    'Order_Value', 
    'Discount_Applied', 
    'Final_Amount', 
    'Delivery_Rating', 
    'Restaurant_Rating', 
    'Profit_Margin'
]


existing_num_cols = [col for col in numeric_cols if col in df.columns]


corr_matrix = df[existing_num_cols].corr()


plt.figure(figsize=(12, 8))
sns.set_theme(style="white")


mask = np.triu(np.ones_like(corr_matrix, dtype=bool))


heatmap = sns.heatmap(
    corr_matrix, 
    mask=mask,
    annot=True, 
    fmt=".2f", 
    cmap='coolwarm', 
    vmin=-1, 
    vmax=1, 
    center=0,
    square=True, 
    linewidths=.5, 
    cbar_kws={"shrink": .8}
)

plt.title('Correlation Heatmap Among Numeric Features', fontsize=14, fontweight='bold', pad=15)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(rotation=0, fontsize=10)

plt.tight_layout()
plt.show()


print("--- Top Correlation Pairs ---")

corr_unstacked = corr_matrix.unstack()
sorted_pairs = corr_unstacked[mask.flatten() == False].abs().sort_values(ascending=False)


distinct_pairs = sorted_pairs[sorted_pairs < 0.9999]

print("Strongest absolute correlations:")
for (feat1, feat2), corr_val in distinct_pairs.head(10).items():
    actual_corr = corr_matrix.loc[feat1, feat2]
    print(f"{feat1:<20} <-> {feat2:<20} : {actual_corr:+.4f}")
```

```python
import pandas as pd
import numpy as np


if all(col in df.columns for col in ['Profit_Margin', 'Final_Amount']):
    
    df['Profit_Margin_Pct'] = np.where(
        df['Final_Amount'] > 0, 
        (df['Profit_Margin'] / df['Final_Amount']) * 100, 
        0.0
    )
elif 'Profit_Margin' in df.columns:
    
    if df['Profit_Margin'].max() <= 1.0:
        df['Profit_Margin_Pct'] = df['Profit_Margin'] * 100
    else:
        df['Profit_Margin_Pct'] = df['Profit_Margin']


if 'Profit_Margin_Pct' in df.columns:
    margin_bins = [-float('inf'), 0, 10, 20, 100]
    margin_labels = ['Loss / Zero (<0%)', 'Low Margin (0-10%)', 'Moderate Margin (10-20%)', 'High Margin (20%+)']
    
    df['Profit_Margin_Tier'] = pd.cut(df['Profit_Margin_Pct'], bins=margin_bins, labels=margin_labels, include_lowest=True)


print("--- Derived Feature: Profit Margin Percentage Summary ---")
print(df['Profit_Margin_Pct'].describe().round(2))

print("\n--- Profit Margin Tiers Distribution ---")
if 'Profit_Margin_Tier' in df.columns:
    print(df['Profit_Margin_Tier'].value_counts(dropna=False))
    print("\nPercentage Share (%):")
    print((df['Profit_Margin_Tier'].value_counts(normalize=True) * 100).round(2))
```

```python
import pandas as pd
import numpy as np

delivery_time_bins = [0, 30, 45, 60, float('inf')]
delivery_time_labels = ['On-Time (<30 mins)', 'Slight Delay (30-45 mins)', 'Moderate Delay (45-60 mins)', 'Severe Delay (>60 mins)']

df['Delivery_Performance_Category'] = pd.cut(
    df['Delivery_Time_Min'], 
    bins=delivery_time_bins, 
    labels=delivery_time_labels, 
    include_lowest=True
)


if 'Estimated_Delivery_Time_Min' in df.columns:
    
    df['Delivery_Delay_Min'] = df['Delivery_Time_Min'] - df['Estimated_Delivery_Time_Min']
    

    conditions = [
        (df['Delivery_Delay_Min'] <= 0),
        (df['Delivery_Delay_Min'] > 0) & (df['Delivery_Delay_Min'] <= 15),
        (df['Delivery_Delay_Min'] > 15) & (df['Delivery_Delay_Min'] <= 30),
        (df['Delivery_Delay_Min'] > 30)
    ]
    sla_labels = ['Early / On-Time', 'Slightly Late (1-15 mins)', 'Late (15-30 mins)', 'Severely Late (>30 mins)']
    
    df['Delivery_SLA_Category'] = np.select(conditions, sla_labels, default='Unknown')



print("--- Derived Feature: Delivery Performance Category Distribution ---")
print(df['Delivery_Performance_Category'].value_counts(dropna=False))

print("\nPercentage Share (%):")
print((df['Delivery_Performance_Category'].value_counts(normalize=True) * 100).round(2))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming df is your cleaned DataFrame containing the online food delivery orders

# --- STEP 1: FILTER CANCELLED ORDERS & AGGREGATE REASONS ---
# Identify cancellation reason and status columns
status_col = 'Order_Status' if 'Order_Status' in df.columns else 'Status'
reason_col = 'Cancellation_Reason' if 'Cancellation_Reason' in df.columns else ('Reason' if 'Reason' in df.columns else 'Cancel_Reason')

# Filter for cancelled orders
cancelled_df = df[df[status_col].astype(str).str.lower() == 'cancelled'].copy()

# Fill missing/unspecified reasons if any exist
cancelled_df[reason_col] = cancelled_df[reason_col].fillna('Unspecified / Other')

# Aggregate cancellation counts, percentage share, and lost revenue
reason_summary = cancelled_df.groupby(reason_col, observed=False).agg(
    Cancellation_Count=('Order_ID', 'count'),
    Total_Lost_Revenue=('Final_Amount', 'sum'),
    Avg_Order_Value=('Final_Amount', 'mean'),
    Avg_Delay_Before_Cancel=('Delivery_Time_Min', 'mean') if 'Delivery_Time_Min' in cancelled_df.columns else ('Order_ID', lambda x: 0.0)
).reset_index()

total_cancellations = reason_summary['Cancellation_Count'].sum()
reason_summary['Share_%'] = ((reason_summary['Cancellation_Count'] / total_cancellations) * 100).round(2)
reason_summary['Total_Lost_Revenue'] = reason_summary['Total_Lost_Revenue'].round(2)
reason_summary['Avg_Order_Value'] = reason_summary['Avg_Order_Value'].round(2)

# Sort by cancellation count
reason_summary = reason_summary.sort_values(by='Cancellation_Count', ascending=False)

# --- STEP 2: VISUALIZATION ---
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot A: Top Cancellation Reasons by Order Count & Share (%)
sns.barplot(
    data=reason_summary, 
    x='Cancellation_Count', 
    y=reason_col, 
    palette='Reds_r', 
    ax=axes[0]
)
axes[0].set_title('Top Reasons for Order Cancellations', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Number of Cancelled Orders', fontsize=10)
axes[0].set_ylabel('Cancellation Reason', fontsize=10)

for p in axes[0].patches:
    width = p.get_width()
    if width > 0:
        pct = (width / total_cancellations) * 100
        axes[0].annotate(f'{int(width):,} ({pct:.1f}%)', 
                         (width, p.get_y() + p.get_height() / 2.), 
                         ha='left', va='center', fontsize=9, xytext=(5, 0), 
                         textcoords='offset points')

# Plot B: Total Lost Revenue by Cancellation Reason
sns.barplot(
    data=reason_summary, 
    x='Total_Lost_Revenue', 
    y=reason_col, 
    palette='Purples_r', 
    ax=axes[1]
)
axes[1].set_title('Total Financial Impact (Lost Revenue) by Reason', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Lost Revenue (₹)', fontsize=10)
axes[1].set_ylabel('')

for p in axes[1].patches:
    width = p.get_width()
    if width > 0:
        axes[1].annotate(f'₹{width:,.0f}', 
                         (width, p.get_y() + p.get_height() / 2.), 
                         ha='left', va='center', fontsize=9, xytext=(5, 0), 
                         textcoords='offset points')

plt.tight_layout()
plt.show()

# --- STEP 3: PRINT EXECUTIVE SUMMARY ---
print("--- CANCELLED ORDERS REASON ANALYSIS ---")
print(f"Total Cancelled Orders: {total_cancellations:,}")
print(f"Total Revenue Lost: ₹{reason_summary['Total_Lost_Revenue'].sum():,.2f}\n")
print(reason_summary.to_string(index=False))
```
